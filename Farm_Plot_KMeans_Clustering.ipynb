```python
# Import necessary libraries

import numpy as np
import matplotlib.pyplot as plt

from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
```

```python
# Step 1: Generate a synthetic dataset
# Create 300 simulated farm plots arranged around 4 groups

X, y_true = make_blobs(
    n_samples=300,
    centers=4,
    cluster_std=2.5,
    random_state=42
)

print("X shape:", X.shape)
print("First 5 farm plots:")
print(X[:5])
```

```python
# View the farm plots before clustering

plt.figure(figsize=(8, 6))

plt.scatter(
    X[:, 0],
    X[:, 1],
    s=50
)

plt.title("Farm Plots Before K-means Clustering")
plt.xlabel("Feature 1 - Simulated Soil pH")
plt.ylabel("Feature 2 - Simulated Moisture Level")

plt.grid(True)
plt.show()
```

```python
# Step 2: Apply K-means clustering

kmeans = KMeans(
    n_clusters=4,
    random_state=40,
    n_init=10
)

# Let K-means study the 300 farm plots
kmeans.fit(X)

# Get the cluster assigned to each farm plot
labels = kmeans.labels_

# Get the center of each cluster
centers = kmeans.cluster_centers_

print("K-means clustering complete!")
print("First 10 cluster assignments:")
print(labels[:10])

print("\nCluster centers:")
print(centers)
```

```python
# Step 3: Visualize the K-means clustering results

plt.figure(figsize=(8, 6))

# Plot the 300 farm plots and color them by cluster
plt.scatter(
    X[:, 0],
    X[:, 1],
    c=labels,
    cmap='viridis',
    s=50,
    label='Farm Plots'
)

# Plot the center of each cluster
plt.scatter(
    centers[:, 0],
    centers[:, 1],
    c='red',
    s=200,
    marker='X',
    label='Cluster Centers'
)

plt.title('K-means Clustering of Farm Plots')
plt.xlabel('Feature 1 - Simulated Soil pH')
plt.ylabel('Feature 2 - Simulated Moisture Level')

plt.legend()
plt.grid(True)

plt.show()
```

```python
# Step 4: Display the clustering results

print("First 10 farm plot cluster assignments:")
print(labels[:10])

print("\nFour cluster centers:")
print(centers)
```
