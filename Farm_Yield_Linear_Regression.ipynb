```python
# Import necessary libraries
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.datasets import make_regression
from mpl_toolkits.mplot3d import Axes3D
```

```python
# Step 1: Generate a synthetic dataset
# We are creating practice farm data with 100 examples

X, y = make_regression(
    n_samples=100,
    n_features=2,
    noise=10,
    random_state=42
)

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
# Make all crop yield values positive

y = (y - np.min(y) + 150) / 2

print("First 10 crop yield values:")
print(y[:10])
```

```python
# Give our two input features meaningful names

soil_ph = X[:, 0]
moisture_level = X[:, 1]

print("First 5 Soil pH values:")
print(soil_ph[:5])

print("First 5 Moisture Level values:")
print(moisture_level[:5])
```

```python
# Step 2: Create and train the Linear Regression model

model = LinearRegression()

model.fit(X, y)

y_pred = model.predict(X)

print("Model training complete!")
print("First 5 actual yields:")
print(y[:5])

print("First 5 predicted yields:")
print(y_pred[:5])
```

```python
# Step 3: Create a 3D graph

fig = plt.figure(figsize=(10, 8))

ax = fig.add_subplot(111, projection='3d')

# Plot the actual crop yields
ax.scatter(
    soil_ph,
    moisture_level,
    y,
    color='blue',
    label='Actual Yields',
    s=50
)

# Plot the predicted crop yields
ax.scatter(
    soil_ph,
    moisture_level,
    y_pred,
    color='red',
    label='Predicted Yields',
    s=50,
    alpha=0.5
)

print("3D scatter plot created!")
```

```python
# Create values across the Soil pH and Moisture ranges
x1_range = np.linspace(soil_ph.min(), soil_ph.max(), 10)
x2_range = np.linspace(moisture_level.min(), moisture_level.max(), 10)

# Create a grid
x1_grid, x2_grid = np.meshgrid(x1_range, x2_range)

# Combine the two grid variables
X_grid = np.c_[x1_grid.ravel(), x2_grid.ravel()]

# Ask our trained model to predict yield across the grid
y_grid = model.predict(X_grid).reshape(x1_grid.shape)

print("Regression grid created!")
```

```python
# Create the final 3D visualization

fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

# Plot actual crop yields
ax.scatter(
    soil_ph,
    moisture_level,
    y,
    color='blue',
    label='Actual Yields',
    s=50
)

# Plot predicted crop yields
ax.scatter(
    soil_ph,
    moisture_level,
    y_pred,
    color='red',
    label='Predicted Yields',
    s=50,
    alpha=0.5
)

# Add the regression plane
ax.plot_surface(
    x1_grid,
    x2_grid,
    y_grid,
    color='green',
    alpha=0.2
)

# Add labels
ax.set_xlabel('Soil pH')
ax.set_ylabel('Moisture Level')
ax.set_zlabel('Crop Yield')

# Add a title
ax.set_title('Crop Yield Prediction Using Linear Regression')

# Add legend
ax.legend()

# Display the graph
plt.show()
```
